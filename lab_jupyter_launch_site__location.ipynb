Next, let's create markers for all launch records.
If a launch was successful `(class=1)`, then we use a green marker and if a launch was failed, we use a red marker `(class=0)`


Note that a launch only happens in one of the four launch sites, which means many launch records will have the exact same coordinate. Marker clusters can be a good way to simplify a map containing many markers having the same coordinate.


Let's first create a `MarkerCluster` object


```python
marker_cluster = MarkerCluster()
```

```python

# Apply a function to check the value of `class` column
# If class=1, marker_color value will be green
# If class=0, marker_color value will be red
spacex_df['marker_color'] = ''
spacex_df.loc[spacex_df['class'] == 0, 'marker_color'] = 'red'
spacex_df.loc[spacex_df['class'] == 1, 'marker_color'] = 'green'
```

```python
# Add Mouse Position to get the coordinate (Lat, Long) for a mouse over on the map
formatter = "function(num) {return L.Util.formatNum(num, 5);};"
mouse_position = MousePosition(
    position='topright',
    separator=' Long: ',
    empty_string='NaN',
    lng_first=False,
    num_digits=20,
    prefix='Lat:',
    lat_formatter=formatter,
    lng_formatter=formatter,
)

site_map.add_child(mouse_position)
```

```python
# Add marker_cluster to current site_map


# for each row in spacex_df data frame
# create a Marker object with its coordinate
# and customize the Marker's icon property to indicate if this launch was successed or failed, 
# e.g., icon=folium.Icon(color='white', icon_color=row['marker_color']

# Add Mouse Position to get the coordinate (Lat, Long) for a mouse over on the map
formatter = "function(num) {return L.Util.formatNum(num, 5);};"
mouse_position = MousePosition(
    position='topright',
    separator=' Long: ',
    empty_string='NaN',
    lng_first=False,
    num_digits=20,
    prefix='Lat:',
    lat_formatter=formatter,
    lng_formatter=formatter,
)

site_map.add_child(mouse_position)
site_map

from math import sin, cos, sqrt, atan2, radians

def calculate_distance(lat1, lon1, lat2, lon2):
    # approximate radius of earth in km
    R = 6373.0

    lat1 = radians(lat1)
    lon1 = radians(lon1)
    lat2 = radians(lat2)
    lon2 = radians(lon2)

    dlon = lon2 - lon1
    dlat = lat2 - lat1

    a = sin(dlat / 2)**2 + cos(lat1) * cos(lat2) * sin(dlon / 2)**2
    c = 2 * atan2(sqrt(a), sqrt(1 - a))

    distance = R * c
    return distance

location = []


# Set marker_color (adapt if your column is different, e.g., based on 'Success?' == True)
spacex_df['marker_color'] = spacex_df['class'].apply(lambda x: 'green' if x == 0 else 'red')

# Check if DataFrame is empty
if spacex_df.empty:
    print("Error: spacex_df is empty! No markers added.")
else:
    print(f"DataFrame has {len(spacex_df)} rows. Proceeding...")

# Create base map (adjust zoom/location as needed)
nasa_coordinate = [29.559684888503615, -95.0830971930759]
location=list(nasa_coordinate)
site_map = folium.Map(nasa_coordinate, zoom_start=5)  # Wider zoom to see clusters

print("Initial map children count:", len(site_map._children))

# Create and add marker cluster
marker_cluster = MarkerCluster().add_to(site_map)
print("After adding cluster, map children count:", len(site_map._children))

# Loop to add markers
added_count = 0
for index, row in spacex_df.iterrows():
    coordinate = [row['Lat'], row['Long']]
    
    marker = folium.Marker(
        coordinate,
        popup=folium.Popup(f"{row['Launch Site']}"),
        icon=folium.Icon(color='white', icon_color=row['marker_color'])
    )
    
    marker_cluster.add_child(marker)
    added_count += 1
    print(f"Added marker {added_count}: {row['Launch Site']} (color: {row['marker_color']})")

print(f"Final map children count: {len(site_map._children)}")
print(f"Cluster has {len(marker_cluster._children)} markers added.")
site_map.add_child(mouse_position)

# Create and add a folium.Marker on your selected closest coastline point on the map
# Display the distance between coastline point and launch site using the icon property 
# for example
coordinate = [28.56231, -80.57734]
coordinates = [[28.56231, -80.57734], [28.56247, -80.56786]]
distance_coastline = calculate_distance(28.56231, -80.57734, 28.56247, -80.56786)
distance_marker = folium.Marker(
    coordinate,
    icon=DivIcon(
        icon_size=(20,20),
        icon_anchor=(0,0),
        html='<div style="font-size: 12; color:#d35400;"><b>%s</b></div>' % "{:10.2f} KM".format(distance_coastline),
       )
    )
locations = list(coordinates)
locations

lines=folium.PolyLine(locations, weight=1)
site_map.add_child(lines)

# For Jupyter/Colab: Display the updated map
if 'IPython' in globals():  # Check if in notebook
    display(site_map)
else:
    # For scripts: Save to HTML and open in browser
    site_map.save('spacex_launch_map.html')
    print("Map saved to 'spacex_launch_map.html' — open it in your browser to view!")
    print("If in a notebook, add 'site_map' at the end of your cell to display inline.")

```
